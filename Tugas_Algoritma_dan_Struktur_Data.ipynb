```python
#Praktikum 1_Tipe Data

#Membuat Variabel
nama="Fajar"
umur=19
berat=54,5
terdaftar=True

#Menampilkan semua nilai variabel dan tipe data
print("Nama:", nama, " Tipe data:",type(nama))
print("Umur:", umur, "Tipe data:", type(umur))
print("Berat:", berat, "Tipe data:", type(berat))
print("Terdaftar:", "Tipe data:", type(terdaftar))

print()

#Mengubah nilai umur menjadi 20
umur=20

#Menampilkan nilai umur yang sesudah diganti nilai umur nya
print("===NIlai Umur Yang Sudah di Ubah===")
print("Nama:", nama, " Tipe data:",type(nama))
print("Umur:", umur, "Tipe data:", type(umur))
print("Berat:", berat, "Tipe data:", type(berat))
print("Terdaftar:", "Tipe data:", type(terdaftar))

```

```python
#Praktikum 2_Operator Aritmatika

#Membuat variabel
harga = 12500
jumlah = 4

#Menghitung total harga
total = harga*jumlah
#Hitung sisa pembagian
sisa_bagi= total % 2000
#Hitung harga diskon
diskon= total - 5000

#Hasil perhitungan
print("===Hasil Perhitungan===")
print("Harga per buku: ", harga)
print("Jumlah buku: ", jumlah)
print("Total harga: Rp", total)
print("Sisa bagi: ", sisa_bagi)
print("After Diskon: Rp", diskon)
```

```python
#Praktikum 3_Operator perbandingan dan logika

print("===Hasil Perhitungan nilai=80 dan Kehadiran=90===")
#Membuat variabel
nilai = 82
kehadiran = 90

#Memeriksa masing-masing syarat
syarat_nilai = nilai >= 75
syarat_kehadiran = kehadiran >= 80

#Menentukan apakah mahasiswa memenuhi kedua syarat
lulus = syarat_nilai and syarat_kehadiran
print("Status Mahasiswa: ", lulus)

print()
#Mengubah nilai menjadi 70
print("===Hasil Perhitungan nilai=70 dan Kehadiran=90 (and)===")
nilai = 70
syarat_nilai = nilai >= 75
lulus = syarat_nilai and syarat_kehadiran
print("Status Mahasiswa: ", lulus)

print()
#Mengubah nilai menjadi 70 (OR)
print("===Hasil Perhitungan nilai=70 dan Kehadiran=90 (OR)===")
nilai = 70
syarat_nilai = nilai >= 75
lulus = syarat_nilai or syarat_kehadiran
print("Status Mahasiswa: ", lulus)

```

```python
#Praktikum 4_Operator Penugasan

#Simpan poin awal dalam variabel poin
poin = 100
print("Nilai awal : ", poin)

#Tambahkan 25 poin menggunakan +=
poin += 25
print("Penambahan (poin += 25) :", poin)

#Kurangi 10 poin menggunakan -=
poin -= 10
print("Pengurangan (poin -= 10) :", poin)

#Kalikan poin dengan 2 menggunakan *=
poin *= 2
print("Perkalian (poin *= 2) :", poin)

#Bagi poin dengan 2 menggunakan /=
poin /= 2
print("Pembagian (poin /= 2) :", poin)
```

```python
#Praktikum 5_Struktur Data Dasar

#Membuat variabel
nilai = [78, 85, 90, 85]
mahasiswa = {"nama": "Nadia", "kelas": "IF-A"}
koordinat = (5, 8)
angka = {2, 4, 4, 6, 8}

#Tampilkan nilai pertama dari list
n = nilai[0]
print(n)

#Tambahkan nilai 95 ke list dan tampilkan list terbaru
nilai = [78, 85, 90, 85, 95]
print("List terbaru :", nilai)

#Tampilkan nama mahasiswa dari dictionary
print("Nama :", (mahasiswa["nama"]))

#Tampilkan koordinat pertama dari tuple
n = koordinat[0]
print(n)

#Tampilkan angka unik dengan sorted(angka)
print(sorted(angka))
```

```python
# Program Rekap Nilai Mahasiswa

# Mahasiswa yang memenuhi syarat
nama_1 = "Sopi"
nilai_tugas_1 = 80
nilai_ujian_1 = 75

# Menghitung rata-rata nilai tugas dan ujian
rata_rata_1 = (nilai_tugas_1 + nilai_ujian_1) / 2

# Memeriksa status kelulusan (lulus jika rata-rata minimal 75)
status_lulus_1 = rata_rata_1 >= 75

print("--- Hasil Rekap Nilai 1 ---")
print("Nama Mahasiswa :", nama_1)
print("Nilai Tugas    :", nilai_tugas_1)
print("Nilai Ujian    :", nilai_ujian_1)
print("Rata-rata      :", rata_rata_1)
print("Status Lulus   :", status_lulus_1)

print("\n" + "="*30 + "\n")

# Mahasiswa yang tidak memenuhi syarat
nama_2 = "ciko"
nilai_tugas_2 = 70
nilai_ujian_2 = 65

# Menghitung rata-rata nilai tugas dan ujian
rata_rata_2 = (nilai_tugas_2 + nilai_ujian_2) / 2

# Memeriksa status kelulusan (lulus jika rata-rata minimal 75)
status_lulus_2 = rata_rata_2 >= 75

print("--- Hasil Rekap Nilai 2 ---")
print("Nama Mahasiswa :", nama_2)
print("Nilai Tugas    :", nilai_tugas_2)
print("Nilai Ujian    :", nilai_ujian_2)
print("Rata-rata      :", rata_rata_2)
print("Status Lulus   :", status_lulus_2)
```
